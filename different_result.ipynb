```python
import pandas as pd

# Load kedua file CSV
file1 = "runs/d_75_15_10_detect/y11n_batch16_epochs10/results.csv"
file2 = "runs/d_70_20_10_detect/y11n_batch16_epochs10/results.csv"

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# Ambil hasil evaluasi terakhir (epoch terakhir)
last_row1 = df1.iloc[-1]
last_row2 = df2.iloc[-1]

# Format hasil untuk dibandingkan
def format_results(row, dataset_name):
    return f"{dataset_name} Summary:\n" \
           f"    Metrics      Precision      Recall      mAP50      mAP50-95\n" \
           f"      all        {row['metrics/precision(B)']:.3f}      {row['metrics/recall(B)']:.3f}      {row['metrics/mAP50(B)']:.3f}      {row['metrics/mAP50-95(B)']:.3f}\n"

# Cetak hasil perbandingan
print(format_results(last_row1, "YOLO11n 75-15-10"))
print(format_results(last_row2, "YOLO11n 70-20-10"))

```
