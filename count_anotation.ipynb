```python
import os
from collections import Counter

# Nama class (urutan harus sesuai dengan urutan ID di file .txt)
class_names = ["nonplastic", "plastic"]

# Path ke dataset
dataset_root = 'dataset_clean_flip'
splits = ['train', 'valid', 'test']

def count_labels(labels_dir):
    counter = Counter()
    if not os.path.exists(labels_dir):
        print(f"[!] Folder tidak ditemukan: {labels_dir}")
        return counter
    for file_name in os.listdir(labels_dir):
        if file_name.endswith('.txt'):
            with open(os.path.join(labels_dir, file_name), 'r') as f:
                for line in f:
                    if line.strip() == '':
                        continue
                    class_id = int(line.strip().split()[0])
                    counter[class_id] += 1
    return counter

# Proses semua split
for split in splits:
    labels_path = os.path.join(dataset_root, split, 'labels')
    print(f"\n📂 Split: {split}")
    counts = count_labels(labels_path)
    total = sum(counts.values())
    print(f"Total anotasi: {total}")
    for i, name in enumerate(class_names):
        print(f" - {name} ({i}): {counts[i]} anotasi")
```
