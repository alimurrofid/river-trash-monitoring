### Menghitung jumlah dan presentasi pembagian dataset

```python
import os

# Path ke dataset dan format gambar yang digunakan
dataset_root = 'datasets/withoutgrogol'
splits = ['train', 'valid', 'test']
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.webp')  # tambahkan jika perlu

# Menyimpan jumlah gambar per split
split_image_counts = {}

# Hitung jumlah file gambar pada folder images setiap split
for split in splits:
    images_path = os.path.join(dataset_root, split, 'images')
    if not os.path.exists(images_path):
        print(f"[!] Folder tidak ditemukan: {images_path}")
        split_image_counts[split] = 0
        continue

    count = len([
        f for f in os.listdir(images_path)
        if f.lower().endswith(image_extensions) and os.path.isfile(os.path.join(images_path, f))
    ])
    split_image_counts[split] = count
    print(f"📂 Split: {split} - {count} gambar")

# Hitung total semua gambar
total_images = sum(split_image_counts.values())

# Tampilkan persentase per split
print("\n📈 Persentase Pembagian Dataset berdasarkan Jumlah Gambar:")
for split in splits:
    count = split_image_counts[split]
    percent = (count / total_images * 100) if total_images > 0 else 0
    print(f" - {split}: {count} gambar ({percent:.2f}%)")

# Tambahkan total keseluruhan data
print(f"\n🧾 Total Keseluruhan Gambar: {total_images}")
```

### Menghitung jumlah dan presentasi anotasi dataset

```python
import os
from collections import Counter

# Nama class (urutan harus sesuai dengan urutan ID di file .txt)
class_names = ["nonplastic", "plastic"]

# Path ke dataset
dataset_root = 'datasets/dataset_clean_flip'
splits = ['train', 'valid', 'test']

# Menyimpan hasil total keseluruhan
overall_counter = Counter()

def count_labels(labels_dir):
    counter = Counter()
    if not os.path.exists(labels_dir):
        print(f"[!] Folder tidak ditemukan: {labels_dir}")
        return counter
    for file_name in os.listdir(labels_dir):
        if file_name.endswith('.txt'):
            with open(os.path.join(labels_dir, file_name), 'r') as f:
                for line in f:
                    if line.strip() == '':
                        continue
                    class_id = int(line.strip().split()[0])
                    counter[class_id] += 1
    return counter

# Proses semua split dan simpan hasilnya
split_counters = {}

for split in splits:
    labels_path = os.path.join(dataset_root, split, 'labels')
    print(f"\n📂 Split: {split}")
    counts = count_labels(labels_path)
    split_counters[split] = counts
    total = sum(counts.values())
    overall_counter += counts

    print(f"Total anotasi: {total}")
    for i, name in enumerate(class_names):
        count = counts[i]
        percent = (count / total * 100) if total > 0 else 0
        print(f" - {name} ({i}): {count} anotasi ({percent:.2f}%)")

# Hitung dan tampilkan distribusi secara keseluruhan
total_overall = sum(overall_counter.values())
print(f"\n📊 Distribusi Keseluruhan (Total: {total_overall} anotasi)")
for i, name in enumerate(class_names):
    count = overall_counter[i]
    percent = (count / total_overall * 100) if total_overall > 0 else 0
    print(f" - {name} ({i}): {count} anotasi ({percent:.2f}%)")

# Tampilkan persentase pembagian dataset berdasarkan jumlah anotasi
print("\n📈 Persentase Pembagian Dataset per Split:")
for split in splits:
    split_total = sum(split_counters[split].values())
    split_percent = (split_total / total_overall * 100) if total_overall > 0 else 0
    print(f" - {split}: {split_total} anotasi ({split_percent:.2f}%)")
```
